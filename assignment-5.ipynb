#### Exercise 3: Reinforcement Learning with PPO (1 + 1: 2 points)

When fine-tuning language models using Reinforcement Learning (such as Proximal Policy Optimization, PPO), we rely on a reward model to evaluate the generated text. To prevent the active policy model from degrading, a Kullback-Leibler (KL) divergence penalty is subtracted from the reward.

The adjusted reward for generating a token $y$ given a prompt $x$ is calculated as:


$$R_{total} = R(x, y) - \beta \log_2 \left( \frac{P_{policy}(y|x)}{P_{ref}(y|x)} \right)$$

Assume you have a toy vocabulary of three tokens: **{Argue, Persuade, Concede}**. For a specific prompt, your models output the following probabilities, and your reward model provides the following scores:

| Token ($y$) | $P_{ref}(y\|x)$ | $P_{policy}(y\|x)$ | $R(x, y)$ |
| --- | --- | --- | --- |
| Argue | 0.25 | 0.50 | 4.0 |
| Persuade | 0.50 | 0.25 | 2.0 |
| Concede | 0.25 | 0.25 | -1.0 |

Assume the KL penalty coefficient is **$\beta = 1.0$**.

a) Calculate the KL penalty term **and** the final adjusted reward $R_{total}$ if the model generates the token **"Argue"**. Show your calculation. 

b) Why is it important to include this KL divergence penalty when fine-tuning with PPO? What would likely happen to the generated text if we removed it (i.e., set $\beta = 0$)? Answer in 2–3 sentences. 

**a)For "Argue"**

**KL penalty term:**

$$\beta \log_2 \left( \frac{P_{policy}(y|x)}{P_{ref}(y|x)} \right) = 1.0 \times \log_2 \left( \frac{0.50}{0.25} \right) = \log_2(2) = 1.0$$

**Adjusted reward:**

$$R_{total} = R(x, y) - \beta \log_2 \left( \frac{P_{policy}(y|x)}{P_{ref}(y|x)} \right) = 4.0 - 1.0 = 3.0$$

---

**b)KL Divergence Penalty**

KL divergence penalty can prevent the model from deviating too far from the reference model.   
Without it, the generated text would weirdly fit the reward model's preference, that would lead to a higher score but loses the linguistic quality.

#### Exercise 5: Decoding Strategies (3 + 1 = 4 points)

To generate a text from a language model, different decoding strategies can be used. Decoding strategies dictate how to choose a word from the outputted probabilities at each step $i$:


$$\prod_{i=1}^{n}{P(w_i|w_1 \cdots w_{i-1})}$$

Here are the decoding strategies used in practice:

* Greedy Decoding: generate at each step the most probable word.
* Beam Search: at each step generate all possible candidates by looking at all the vocabulary and keep the $k$ most probable sequence candidates.
* Random Sampling: sample the words according to the probability distribution of $P(w_i|w_1 \cdots w_{i-1})$.
* Top-k sampling: truncate the probability distribution $P(w_i|w_1 \cdots w_{i-1})$ to a certain length of $k$ and redistribute the truncated probability mass to these selected tokens.

a) Discuss the advantages and problems of each decoding strategy specifically in the context of generating counterarguments. Consider in your discussion aspects like development and computational cost, adequacy (how well the generated text logically counters the premise), creativeness, and diversity of the generated text. To support the discussion, we provide hypothetical generated outputs for each decoding strategy given the prompt: *Generate a counterargument. Premise: Nuclear energy should be phased out because it produces radioactive waste that is dangerous to store.*

* Greedy: *On the other hand, nuclear energy is a reliable source of energy. Nuclear energy is a reliable source of energy. Nuclear energy is a reliable*
* Beam Search: *However, nuclear energy provides a stable base-load power supply that solar and wind power cannot match. Therefore, nuclear power is necessary.*
* Random Sampling: *Conversely, atomic crystals vibrate rapidly powering ocean submarines which offsets the radioactive sludge dilemma.*
* Top-k sampling: *While waste storage is a challenge, modern reactor designs can actually recycle this waste to generate even more electricity, making it highly efficient.*

b) Which of the decoding strategies fits better for applications such as machine translation and which fits more for generating argumentative conclusions? Explain your reasoning in 1-2 sentences.

**a)**

**Greedy Decoding**

*Advantages:*  
**Very low computational cost** since we only evaluate one token per step, making it **easy to develop and deploy**. It can still produce **topically adequate output** because we are choosing the highest-probability token to address the premise.

*Problems:*  
**Creativity and diversity are not good** because every run produces the same output and explores no alternative argumentative angles.(as seen: "Nuclear energy is a reliable source of energy" repeated)

---

**Beam Search**

*Advantages:*  
Beam search consistently produces **grammatically correct and logically coherent** counterarguments with **high adequacy** (as in the example: a well-structured rebuttal about base-load power supply). Development **cost is manageable** and the **output is deterministic and reproducible**.

*Problems:*  
Since we are only considering the top k choices, all k beams tend to converge toward the same high-probability sequences, resulting in **low diversity and low creativity**, limiting the range of argumentative perspectives.

---

**Random Sampling**

*Advantages:*  
**High diversity and creativity**, sampling from the full distribution allows unexpected and novel argumentative angles to emerge. Also **computationally cheap**, comparable to greedy decoding.

*Problems:*  
**Adequacy is very low**. Since low-probability tokens can always be selected, the output can become incoherent to the topic (as in the example: "atomic crystals vibrate rapidly powering ocean submarines"). **Development also becomes harder** since output quality is unpredictable and difficult to control.

---

**Top-k Sampling**

*Advantages:*  
Filters out incoherent choices while preserving meaningful variation. The result is **high adequacy** combined with **meaningful creativity and diversity** (as seen in the example: which produces a novel and logically sound rebuttal about waste recycling.) **Computational cost is low**

*Problems:*  
it's tricky to choose the value of k, while **outputs are nondeterministic**, making exact reproduction difficult.

---

**b)**

**Machine translation: Beam Search.** 
Translation needs to stick to the original content, Beam Search can produce a stable text without being too rigid.

**Argumentative conclusion generation: Top-k sampling.** 
Generate argumentative conclusion needs creativity, top-k sampling might provide aguments from different angles, while avoid generating incoherent text.
