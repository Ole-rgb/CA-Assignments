The task says class I and class O specifically, so the gold-B token (*emails*) is treated as "not-I and not-O" wherever I am counting (so predicting I on *emails* counts as a false positive for class I).

**Class I:**

- TP = tokens where gold=I and pred=I → rows 10..21 → **12**
- FP = pred=I but gold≠I → rows 6, 7, 8 (gold O) plus row 9 (gold B) → **4**
- FN = gold=I but pred≠I → row 22 (pred O) → **1**

$$P_I = \frac{12}{12+4} = \frac{12}{16} = 0.75$$
$$R_I = \frac{12}{12+1} = \frac{12}{13} \approx 0.92$$

**Class O:**

- TP = gold=O and pred=O → rows 1..4 → **4**
- FP = pred=O but gold≠O → row 22 (gold I) → **1**
- FN = gold=O but pred≠O → row 5 (pred B), rows 6, 7, 8 (pred I) → **4**

$$P_O = \frac{4}{4+1} = 0.80, \qquad R_O = \frac{4}{4+4} = 0.50$$

The algorithm essentially starts the span 4 tokens too early (at *think* instead of *emails*) and stops 1 token too early (drops *technology*). That early-start kills the O recall.

**b)** Both algorithms have the same precision+recall sum:

- $a_1$: 0.82 + 0.98 = 1.80
- $a_2$: 0.95 + 0.85 = 1.80

Since $F_1 = \frac{2 PR}{P+R}$, with $P+R$ fixed the score scales with the product $P\cdot R$. The product is largest when $P$ and $R$ are equal, and gets smaller the further apart they are. $a_2$ has $|P-R| = 0.10$ while $a_1$ has $|P-R| = 0.16$, so **$a_2$** gets the higher $F_1$ without doing the full calculation.

Social Bias can affect computational argumentation approaches, for example, an argument generation model can generate arguments that prefer a specific social group. WEAT (Word Embedding Association Test) is an evaluation metric used to measure bias in word embeddings. It assesses the association of word sets with attribute sets to quantify biases such as gender or racial bias in word embeddings. The test calculates the differential association of two sets of target words with two sets of attribute words. For example, it can measure the association of male and female names with stereotypical male/female jobs.

The equation for the Word Embedding Association Test (WEAT) score is as follows:

$$
S(X, Y, A, B) = \sum_{x \in X}s(x, A, B) - \sum_{y \in Y} s(y, A, B)
$$

Where:
* X and Y are two sets of target words.
* A and B are two sets of attribute words.
* s(w, A, B) is a differential association of a word (w) with the two sets of attribute words.

The differential association s(w, A, B) is calculated as the mean of the differences in the cosine similarities of w with all words in A and B:

$$
s(w, A, B) = mean_{a \in A} cos(w, a) - mean_{b \in B} cos(w, b)
$$

A WEAT score of $0$ means no bias, the higher the _absolute_ score the more prominent the bias.
