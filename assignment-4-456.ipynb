#### Exercise 4: Evaluation (1.5 + 0.5 : 2 points) 
You were asked to develop a regression model to assess the effectiveness of a news article to a numerical value in the range $[0,3]$. You are given 1000 news articles on 10 topics where each news article is labeled with 0 (not effective), 1 (somehow effective), 2 (moderately effective), and 3 (very effective). You were also given which authors wrote each news article among ten authors who vary in terms of writing experience, style, and popularity. Also, each article is labeled with pro or con the topic it addresses. Your first steps are to split the dataset into a training and test sets and then to choose a proper evaluation metric for an experimental setup.  

a) Which of the following statements are correct and which are incorrect regarding an appropriate experimental setup? Explain your answer in one or two sentences.  
1. The training set should contain all news articles whose stance is pro, while the test set should contain all news articles that are con. 
2. The training set should contain all news articles that are labeled as very effective or moderately effective, while the test set should contain news articles that are labeled as somehow effective or not effective. 
3. To help the model learn general effectiveness, it is helpful to ensure the training set contains a diverse mix of all ten authors, and the test set should likewise be artificially balanced to contain exactly equal numbers of articles from each author.

b) Which metric of the following will you choose between Mean Squared Error (MSE) and Mean Absolute Error (MAE)? Explain your choice, considering that most of the news articles in the dataset are heavily polarized (labeled as either very effective or not effective).

**a) Correctness of statements:**

1. **Incorrect.** Splitting by stance puts all pro articles in training and all con in testing (or vice versa). The model never sees one stance during training, so at test time it faces a distribution it has never encountered and cannot be expected to generalize to it.

2. **Incorrect.** Splitting by effectiveness label means whole target classes are absent from training. The model learns nothing about, e.g., low-effectiveness articles and is then evaluated only on them, so the evaluation measures nothing meaningful about the model's actual ability.

3. **Incorrect.** Ensuring the *training* set covers all ten authors is good practice and helps generalization. But *artificially balancing the test set* by author is wrong: a test set must be an unbiased sample of the deployment distribution. Forcing equal author counts changes the prior over authors, so the resulting error estimate no longer reflects real-world performance — it over-weights rare authors and under-weights prolific ones.

**b) MSE vs. MAE:**

The choice depends on how we want errors to be penalized, and we should state that assumption rather than asserting one metric blindly.

- **MSE** penalizes large errors quadratically. With a heavily polarized {0, 3} distribution, this forces the model to actually separate the two poles, because a model that just predicts the mean (~1.5) is punished hard. The downside is that a few large misses dominate the score.
- **MAE** weights every point of error equally and is more robust to occasional large gaps, but it is easier to game by regressing toward the mean.

If each effectiveness point matters equally and we do not want a handful of large misses to dominate the metric, **MAE** is the better reported metric. If large misses are genuinely costly for the application, **MSE** is preferable. Given the task framing (a simple effectiveness score where every level is equally important), MAE gives the more interpretable average error, but the key point is matching the metric to the cost of errors.

#### Exercise 5: Fallacy Detection (2 Points)

In the lecture on Argument Quality Assessment, we established that a high-quality argument must be logically sound, meaning its premises must be acceptable, relevant, and provide sufficient support for the conclusion. Arguments that fail these criteria often rely on fallacies (as briefly introduced in Part III)—flaws in reasoning that manipulate rhetorical emotion or distract the reader to hide a lack of logical substance.

Imagine you are working as a computational linguist for a widely used live-streaming debate platform. Because many of the discussed topics are highly controversial, discussions frequently derail into these low-quality, fallacious arguments. To keep the overall quality of discussion high, you are tasked with creating a classifier that flags logically flawed arguments for moderators. You extract the following set of common fallacies from a research article to create your training guidelines:

- Ad Hominem: When the argument becomes an excessive personal attack on the arguer rather than addressing their position.

- Appeal to Emotion: The unessential loading of the argument with emotional language to exploit the audience's emotional instinct rather than relying on logic.

- Appeal to Authority: It occurs when the arguer relies on the endorsement of an authority figure or a group consensus without providing sufficient evidence. It may also involve the citation of non-experts.

- Slippery Slope: This fallacy implies that an improbable, exaggerated, or extreme consequence will inevitably result from a relatively minor initial action.

- False Cause: The misinterpretation of a simple correlation between two events as direct causation.

- Slogan: A brief and striking phrase used to provoke excitement in the audience, often accompanied by "argument by repetition."

Task: Label the following four texts with the single most appropriate fallacy from the list above. Explain your choice for each in 1–2 sentences.

1. "Teacher: If I let you go to the bathroom, pretty soon everybody will want to go to the bathroom, and then no one will be left in the classroom to learn."

2. "That candidate wants to raise the minimum wage, but they aren't even smart enough to run a business."

3. "They cannot take her to jail even if she did rob the bank and take millions of pesos. She is already too old and weak to be locked up. That would be cruel."

4. "My child was diagnosed with autism shortly after receiving vaccinations. The vaccines are clearly to blame."

1. **Slippery Slope.** The teacher claims one student leaving will inevitably cascade into the whole class leaving — an exaggerated, improbable chain of consequences from a minor initial action.

2. **Ad Hominem.** Instead of engaging with the minimum-wage policy, the speaker attacks the candidate's personal intelligence and business competence, i.e., the arguer rather than the argument.

3. **Appeal to Emotion.** The argument concedes she may be guilty, then sidesteps the legal question by exploiting pity over her age and frailty rather than offering a legal reason. (It borders on special pleading, but Appeal to Emotion is the fitting label from the given list because the move is to provoke sympathy.)

4. **False Cause.** The argument reads temporal succession (vaccination, then diagnosis) as direct causation — the classic *post hoc ergo propter hoc* error, mistaking correlation in time for cause.

For each scenario, identify which category it belongs to and justify your decision in 1–2 sentences.  

- Scenario A: You are building a live dashboard for debate moderators that must simultaneously output scores for Clarity, Credibility, and Emotional Appeal for every argument presented. You care about all three outputs equally. 
- Scenario B: Your primary goal is to classify the Overall Argument Quality of an essay. However, your labeled dataset for this is very small. You do have access to a massive, separate dataset annotated for Toxicity Detection (a completely unrelated task focused on detecting profanity and abuse).
- Scenario C: You want to build a highly accurate classifier for Social Bias Detection (your primary task), and you have access to a large dataset for Toxicity Detection. These tasks are distinct but highly similar in their linguistic features.  

b) Hierarchical Architecture Design

The lecture introduces the "Hierarchical Multitask BERT" architecture, which evaluates subdimensions and uses them to inform the overall quality regressor. Based on the complete argument quality taxonomy presented in the lecture (which contains exactly 15 dimensions), draw or describe an extended 2-stage hierarchical architecture. Your architecture must explicitly show how the 11 base dimensions map into the 3 mid-level dimensions, and how those map into the final overall quality score.

**a) Task Scenarios:**

- **Scenario A -> Joint multitask learning only.** All three outputs (Clarity, Credibility, Emotional Appeal) must be produced simultaneously for every argument in real time, which is exactly the setting joint MTL targets: one shared encoder feeding three prediction heads. Continual learning does not fit, because the tasks are parallel outputs, not sequential prerequisites where one must be learned before the next.

- **Scenario B -> Neither.** Toxicity Detection (profanity/abuse) and Overall Argument Quality share little linguistic signal. A shared encoder would pull representations toward toxicity cues that are largely irrelevant to quality, risking negative transfer, so neither joint nor continual MTL can be relied on to help the primary task.

- **Scenario C -> Both are useful.** Social Bias Detection and Toxicity Detection are distinct but linguistically similar, so a shared encoder benefits both (joint MTL). Equally, pre-training on the large Toxicity dataset and then fine-tuning on Social Bias Detection (continual MTL) is sensible given the related features and the value of the large auxiliary dataset.

**b) Hierarchical Architecture (2-stage):**

The lecture's taxonomy (Wachsmuth et al.) groups **11 base dimensions** under **3 mid-level dimensions**, which feed the overall quality regressor:

```
Stage 1 - 11 base dimensions -> 3 mid-level dimensions
(each base dimension has its own BERT-based prediction head)

COGENCY (mid-level):
  [ Local Acceptability ] -+
  [ Local Relevance     ] -+--> COGENCY score
  [ Local Sufficiency   ] -+

EFFECTIVENESS (mid-level):
  [ Credibility         ] -+
  [ Emotional Appeal    ] -+
  [ Clarity             ] -+--> EFFECTIVENESS score
  [ Appropriateness     ] -+
  [ Arrangement         ] -+

REASONABLENESS (mid-level):
  [ Global Acceptability] -+
  [ Global Relevance    ] -+--> REASONABLENESS score
  [ Global Sufficiency  ] -+

Stage 2 - 3 mid-level dimensions -> Overall Quality

  COGENCY score        -+
  EFFECTIVENESS score  -+--> Overall Argument Quality (regression head)
  REASONABLENESS score -+
```

Counting: Cogency has 3 base dimensions, Effectiveness has 5, Reasonableness has 3, for **11 base dimensions** in total. Together with the 3 mid-level dimensions and the 1 overall score this gives the 15-node taxonomy. In the network, each base dimension is predicted by its own head on the shared encoder; the 3 mid-level scores are then combined and passed to a final regression head that outputs overall quality in [0, 1].
