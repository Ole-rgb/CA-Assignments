While developing an argument summarization model (a synthesis task), you obtain the following performance scores across a 5-fold cross-validation. Here, $bleu$ represents your model’s effectiveness, and $bleu'$ represents the baseline's effectiveness. The $bleu$ score calculates the quality of machine-generated text using the count of common tokens in the generated text that occur in the ground-truth texts (usually, a generated text is compared against multiple candidates). You will learn about the formula of the bleu score later in the lecture, it is not needed here.

$$bleu = [0.42, 0.42, 0.42, 0.43, 0.45]$$
$$bleu' = [0.40, 0.42, 0.41, 0.42, 0.44]$$

You want to show that the two models perform similarly. Since your approach is a variation of the baseline model, a dependent t-test for paired samples should be applied. Assuming that the scores are normally distributed, use a dependent t-test for paired samples with a significance level $\alpha$ of 0.05 to find out whether the performance of the model is significantly better than the baseline. While calculating the $t$-score, please always round your decimals to four decimal points. To perform the test, do the following:

#### Instructions

a) Decide whether the test is one-tailed or two-tailed.

b) Formulate the research hypothesis $H$ and the null hypothesis $H_0$.

c) Calculate the $t$-score.

d) Look up the $p$-value for the right degree of freedom at https://www.sjsu.edu/faculty/gerstman/StatPrimer/t-table.pdf

e) Decide whether $H_0$ can be rejected. Explain your decision in 1-2 sentences.

Below, we give an example. If needed, you can find more details on dependent $t$-tests online.

#### Example

Assume that you want to show that a classifier significantly outperforms a baseline in a 5-fold cross-validation setting with the following scores (again multiplied by 100). $Accuracy$ is the scores of the model and $accuracy'$ stands for the scores of the baseline:

$$ accuracy' = [71, 70.5, 71.8, 72, 70.3]$$
$$ accuracy = [73, 74.2, 72.3, 73.3, 74.1]$$

a) Type of test: 

Since we assume the model scores to be larger than the baseline, we perform a one-tailed paired t-test. We take a significance level $\alpha$ of 0.05.

b) Formulating hypotheses:

$H$: the mean of the scores of the model is significantly larger than the mean of the baseline's scores.

$H_0$: the mean of the scores of the model is not significantly larger than the baseline's scores.

c) Calculate a $t$-score as follows:

$$t =\frac{\bar{D}-\mu_{D}}{\frac{s_{D}}{\sqrt{n}}}$$

where

* $\bar{D}$ is the mean distance between the model and the baseline's scores. 
* $\mu_{D}$ is the expected difference between the population means. 
* $s_{D}$ is the standard deviation of the scores distances.
* $n$ is the count of scores which is five here.

$\mu_{D}$ is considered zero since the expected difference of the population mean when the null hypothesis applies is zero.    

$$
\bar{D} = \frac{\sum_{1}^{n}{(accuracy_i - accuracy'_i)}}{n} = 2.26 \\
s_{D}=\sqrt{\frac{\sum_{1}^{n}{((accuracy_i - accuracy'_i)-\bar{D})²}}{n-1}} = 1.46 \\
t = \frac{2.26}{\frac{1.46}{ \sqrt{5}}} = 3.46 \\
$$

d) Look up $p$-value for the right degree of freedom:

Calculating the degree of freedom. The degree of freedom is the number of values that can vary in our sample. Since we consider differences between paired scores, the degrees of freedom are n - 1, which is 4 in this case. 

Looking up the $p$-value in the table [t-table](https://www.sjsu.edu/faculty/gerstman/StatPrimer/t-table.pdf) by finding the largest $ t$-value in the table that is smaller than $t$, which is in this case a $p$-value of $0.025$ for the $t$-value $2.776$.


e) Accept or reject $H_0$.

In this case we reject $H_0$ since 0.025 ($p$-value) $\le$ 0.05 ($\alpha$)

#### Exercise 6 (3 points, 0.5 per segment)

The Toulmin model provides a fine-grained way to capture the internal structure of an argument by assigning specific roles to different text units. Read the following argument regarding university tuition and map the segments (1)-(6) to their corresponding Toulmin unit types (Claim, Facts, Warrant, Backing, Rebuttal, or Qualifier). Justify each answer in 1-2 sentences.

"(1) University tuition should be free for all students because (2) recent economic studies show that a highly educated workforce increases a country's GDP significantly. (3) Generally, investments in human capital lead to long-term economic growth, (4) as established by standard macroeconomic theories. (5) It is likely that this policy would reduce the wealth gap, (6) unless the administration costs of such a program outweigh the economic benefits."


1. Claim - "University tuition should be free" is the core conclusion of the argumentation.Everything else either supports it or qualifies it.
2. Facts - Evidence is given that supports the claim. The economic studies showing the GDP effect are the empirical evidence given directly for the claim. The word "because" marks them as grounds.
3. Warrant - If this holds the facts support the claim.
"investments in human capital lead to long-term economic growth" is the general rule that licenses the step from the GDP evidence (2) to the claim (1). It explains *why* the data supports the claim.
4. Backing - This general statement supports the preceeding warrant.
"as established by standard macroeconomic theories" doesn't support the claim directly, it supports the warrant in (3) by pointing to the theoretical authority behind it. That is exactly what backing does.
5. Qualifier - "It is likely" acknowledges uncertainty in the argument.
"it is likely" weakens the strength of the statement. It is not new evidence and not an objection, it just says how strongly the claim holds, which is the role of a qualifier.
6. Rebuttal - Under this condition the claim does not hold.
"unless the administration costs ... outweigh the economic benefits" introduces a condition under which the claim would not hold. The word "unless" explicitly marks an exception, which is the function of a rebuttal.
